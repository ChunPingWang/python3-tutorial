## 5.2 Python 的四種基本容器

| 容器 | 長相 | 有順序 | 可修改 | 有重複 | 什麼時候用 |
| --- | --- | --- | --- | --- | --- |
| **list** 串列 | `[65, 45, 120]` | ✔ | ✔ | ✔ | 一串同類型的東西 |
| **tuple** 元組 | `(25.03, 121.5)` | ✔ | ✖ | ✔ | 一組固定不變的值 |
| **dict** 字典 | `{"珍奶": 65}` | ✔ | ✔ | 鍵不重複 | **查表** |
| **set** 集合 | `{"珍奶", "紅茶"}` | ✖ | ✔ | ✖ | **去重複**、判斷有沒有 |

> 🔍 **選哪一個？問兩個問題：**
> 1. 我要「一串」還是「一張對照表」？→ 串列 / 字典
> 2. 重複的要留著嗎？→ 要留用串列，不要留用集合

## 5.4 💡 list：一串東西

````python
訂單 = [65, 45, 120]

print(訂單[0],  "← 第一個（索引從 0 開始）")
print(訂單[-1], "← 最後一個（負數從後面數）")
print(訂單[0:2], "← 切片，從 0 到 2（不含 2）")
print(訂單[:2], 訂單[1:], "← 省略代表從頭/到尾")
````

````python
訂單 = [65, 45, 120]
訂單.append(90)          # 尾端加
print("append :", 訂單)
訂單.insert(0, 30)       # 插到最前面
print("insert :", 訂單)
訂單.remove(45)          # 移除「值是 45」的第一個
print("remove :", 訂單)
最後一個 = 訂單.pop()     # 取出並移除最後一個
print("pop    :", 訂單, "取出了", 最後一個)
print("長度   :", len(訂單), "｜65 在裡面嗎：", 65 in 訂單)
````

````python
# ⚠️ 索引超出範圍
try:
    訂單[len(訂單)]
except IndexError as 錯誤:
    print("🔴 IndexError：", 錯誤)
    print("   最後一個的索引是 len-1，不是 len")
````

### ⚠️ `.sort()` 回傳 None！

````python
訂單 = [65, 45, 120]

結果 = 訂單.sort()
print("結果 =", 結果, "← 是 None，不是排序後的清單！")
print("訂單 =", 訂單, "← 原清單已經被改掉了")

訂單 = [65, 45, 120]
新清單 = sorted(訂單)
print("sorted() 回傳新清單：", 新清單, "｜原清單不動：", 訂單)
````

**Python 的慣例：會改動原物件的方法，通常回傳 `None`。**
這是初學者最常問的「為什麼我的清單變成 None」的原因。

### ⚠️⚠️ 最重要的坑：清單是「共用」的

````python
a = [1, 2, 3]
b = a                # 這不是複製！b 和 a 是同一個清單的兩個名字
b.append(4)
print("a =", a, "← a 也被改了！")
print("a is b ?", a is b)
````

````python
a = [1, 2, 3]
b = a.copy()         # 真正的複製（或 a[:] 或 list(a)）
b.append(4)
print("a =", a, "｜b =", b, "｜a is b ?", a is b)
````

> 🔍 **為什麼？** 因為變數是「名牌」（第 1 章）。`b = a` 是把 `b` 這張名牌貼到同一個清單上。
> 字串和數字沒這問題，因為它們不可變。
>
> **凡是 list / dict / set 的指派，都要問自己：我要同一個，還是要一份新的？**

### 串列推導式（list comprehension）

````python
訂單 = [65, 45, 120]

# 😩 一般寫法
含稅清單 = []
for 金額 in 訂單:
    含稅清單.append(round(金額 * 1.05))

# 🙂 推導式：一行搞定
含稅推導 = [round(金額 * 1.05) for 金額 in 訂單]

print(含稅清單)
print(含稅推導)
assert 含稅清單 == 含稅推導

# 加篩選
大額 = [金額 for 金額 in 訂單 if 金額 > 100]
print("大額訂單：", 大額)
````

讀法：**`[要什麼 for 每個東西 in 哪裡 if 條件]`**

⚠️ 不要寫超過一行看不懂的推導式。巢狀兩層以上就改回 `for` 迴圈，**可讀性優先**。

## 5.5 💡 dict：查表 —— 阿宏問題的答案

````python
單價表 = {
    "珍珠奶茶": 65,
    "冬瓜檸檬": 45,
    "紅茶": 35,
}

print(單價表["珍珠奶茶"])       # 查表
單價表["椰果奶茶"] = 60         # 新增
單價表["紅茶"] = 38            # 鍵已存在 → 修改
print(單價表)
print("珍珠奶茶" in 單價表, len(單價表))
````

````python
# ⚠️ 查不到的鍵會爆炸
try:
    單價表["綠茶"]
except KeyError as 錯誤:
    print("🔴 KeyError：", 錯誤)

# 三種安全查法
print(單價表.get("綠茶"))         # None
print(單價表.get("綠茶", 0))      # 指定預設值 ← 最常用
print("綠茶" in 單價表)
````

> 🔍 **什麼時候用 `[]`，什麼時候用 `.get()`？**
>
> - 用 `[]`：這個鍵**一定要存在**，不存在就是 bug，我要它立刻爆給我看。
> - 用 `.get()`：這個鍵**可能不存在**，那是正常情況，給預設值就好。
>
> 別無腦一律用 `.get()` —— 那會把真正的 bug 藏起來，變成後面某處莫名其妙的 `None`。

````python
# 走訪字典
for 品名, 價格 in 單價表.items():          # 最常用
    print(f"{品名:<8}{價格:>4}")

print("\n所有鍵：", list(單價表.keys()))
print("所有值：", list(單價表.values()))
print("總和  ：", sum(單價表.values()))
````

### 巢狀字典：阿宏的完整菜單

````python
菜單 = {
    "珍珠奶茶": {"中杯": 55, "大杯": 65},
    "冬瓜檸檬": {"中杯": 40, "大杯": 45},
    "紅茶": {"中杯": 30, "大杯": 35},
}

print(菜單["珍珠奶茶"]["大杯"])


def 查價(品名, 杯型):
    return 菜單[品名][杯型]


assert 查價("珍珠奶茶", "大杯") == 65
assert 查價("珍珠奶茶", "中杯") == 55
print("🟢 24 個變數 → 一張表。新增品項只要多一行，不用改邏輯。")
````

> 🔍 **字典為什麼快？**
> 字典用雜湊表（hash table）實作，查一筆的時間**和資料量無關** —— 1 萬筆和 10 筆一樣快。
> 用清單搜尋要一筆一筆比對，1 萬筆就慢 1000 倍。
>
> **第 4 章說的「巢狀迴圈太慢」，解法通常就是先建一個字典。**

````python
# 親眼比較一下（不用看懂 timeit，看數字就好）
import timeit

大清單 = list(range(100_000))
大集合 = set(大清單)

清單時間 = timeit.timeit(lambda: 99_999 in 大清單, number=100)
集合時間 = timeit.timeit(lambda: 99_999 in 大集合, number=100)
print(f"在 10 萬筆裡找東西，做 100 次：")
print(f"  清單：{清單時間:.4f} 秒")
print(f"  集合：{集合時間:.6f} 秒  ← 快了幾千倍")
````

## 5.6 💡 set：去重複與集合運算

````python
今日訂單 = ["珍珠奶茶", "冬瓜檸檬", "珍珠奶茶", "紅茶"]

賣出品項 = set(今日訂單)
print(賣出品項, "← 重複自動消失")
print("賣出", len(賣出品項), "種")
````

````python
菜單品項 = {"珍珠奶茶", "冬瓜檸檬", "紅茶", "綠茶"}
今日賣出 = {"珍珠奶茶", "冬瓜檸檬", "紅茶"}

print("滯銷品（差集 -）：", 菜單品項 - 今日賣出)
print("都有（交集 &）  ：", 菜單品項 & 今日賣出)
print("全部（聯集 |）  ：", 菜單品項 | 今日賣出)
````

**一行找出滯銷品，不用寫迴圈。**

⚠️ 空集合要寫 `set()`，不是 `{}`（`{}` 是空字典）。

## 5.7 💡 tuple：不可修改的一組值

````python
一筆訂單 = ("珍珠奶茶", "大杯", 65)
品名, 杯型, 價格 = 一筆訂單          # 拆包 unpacking
print(品名, 杯型, 價格)

try:
    一筆訂單[0] = "紅茶"
except TypeError as 錯誤:
    print("🔴 TypeError：", 錯誤, "← tuple 不能改")
````

tuple 和 list 幾乎一樣，差別只有不能修改。那為什麼要用它？

1. **表達「這組值不該被改」** —— 座標、一筆紀錄
2. **可以當字典的鍵**（list 不行，會變的東西不能當鍵）
3. **函式回傳多個值時，Python 自動用 tuple**

````python
def 統計(訂單):
    return sum(訂單), max(訂單), len(訂單)      # 其實回傳一個 tuple


結果 = 統計([65, 45, 120])
print("原樣：", 結果, type(結果).__name__)

總額, 最大, 筆數 = 統計([65, 45, 120])           # 自動拆包
print(f"{總額=} {最大=} {筆數=}")
````

````python
# tuple 當字典的鍵：記錄每個 (品項, 杯型) 組合賣了幾杯
銷量 = {
    ("珍珠奶茶", "大杯"): 32,
    ("珍珠奶茶", "中杯"): 18,
}
print(銷量[("珍珠奶茶", "大杯")])

# list 不能當鍵
try:
    {["珍珠奶茶", "大杯"]: 32}
except TypeError as 錯誤:
    print("🔴 TypeError：", 錯誤)
````

## 5.8 🟢 讓測試變綠 + 統計熱銷

````python
from collections import Counter


def 品項數量(菜單):
    return len(菜單)


def 賣出品項(訂單):
    return set(訂單)


def 熱銷品項(訂單):
    return Counter(訂單).most_common(1)[0][0]


今日訂單 = ["珍珠奶茶", "冬瓜檸檬", "珍珠奶茶", "紅茶"]

assert 查價("珍珠奶茶", "大杯") == 65
assert 查價("珍珠奶茶", "中杯") == 55
assert 品項數量(菜單) == 3
assert 賣出品項(今日訂單) == {"珍珠奶茶", "冬瓜檸檬", "紅茶"}
assert 熱銷品項(今日訂單) == "珍珠奶茶"
print("🟢 全綠")
````

````python
# Counter 是標準庫送你的計數器（第 12 章詳談）
print(Counter(今日訂單))
print(Counter(今日訂單).most_common(2))

# 手寫版本（理解原理用）：.get 預設值的經典用法
次數 = {}
for 品項 in 今日訂單:
    次數[品項] = 次數.get(品項, 0) + 1
print(次數)
assert 次數 == dict(Counter(今日訂單))
print("🟢 手寫版和 Counter 結果相同")
````
