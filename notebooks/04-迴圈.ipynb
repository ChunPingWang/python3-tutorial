# 第 4 章：迴圈 —— 把重複的事交給電腦

> 🎬 **情境**
> 打烊了。阿宏把訂單本丟給你：「一共 187 筆。算今天賣多少、哪一筆最貴、有幾筆是會員。
> 喔對，明天開始每天都要。」

你昨天處理一筆要寫 3 行，今天要寫 561 行，明天再 561 行。

## 4.1 😩 土法煉鋼：複製貼上

````python
總額 = 0
總額 = 總額 + 65
總額 = 總額 + 45
總額 = 總額 + 65
總額 = 總額 + 55
# ...還有 183 行
print(總額, "（才做了 4 筆…）")
````

問題不只是累：

1. **會漏、會重複**，而且自己看不出來
2. **明天的資料不一樣**，561 行要全部重寫
3. **阿宏改規則**（例如會員不計入），要改 187 個地方

> **只要你在複製貼上程式碼，就代表你需要迴圈。**

## 4.2 🧪 先寫測試

````python
今日訂單 = [65, 45, 65, 55, 120, 90]      # 先用 6 筆代表 187 筆

try:
    assert 總營收(今日訂單) == 440
    assert 最貴一筆(今日訂單) == 120
except NameError as 錯誤:
    print("🔴 紅燈：", 錯誤)
````

> 🔍 **為什麼測試只用 6 筆，不用 187 筆？**
> 因為測試要**跑得快、看得懂、壞了能馬上定位**。6 筆你可以心算驗證，187 筆你不行。
> **好的測試資料是「最小但涵蓋所有情況」的資料**，不是真實資料的複製品。
> 187 筆是拿來驗收的，6 筆是拿來開發的。

## 4.3 💡 for：對每一個東西做一次

````python
for 金額 in 今日訂單:
    print(金額)
````

讀法：**「對於 `今日訂單` 裡的每一個東西，把它叫做 `金額`，然後做底下的事。」**

````python
總額 = 0
for 金額 in 今日訂單:
    總額 += 金額           # 等同 總額 = 總額 + 金額
print(總額)
````

6 筆是這 3 行，187 筆也是這 3 行，10000 筆還是這 3 行。**這就是迴圈的意義。**

### 累加器模式：三步驟

```
總額 = 0          ← 1. 迴圈「外面」準備起始值
for x in 清單:
    總額 += x     ← 2. 迴圈「裡面」更新
print(總額)       ← 3. 迴圈「外面」使用結果
```

````python
# ⚠️ 最常見的錯誤 1：把起始值放進迴圈裡
總額 = 0
for 金額 in 今日訂單:
    總額 = 0          # ✖ 每一圈都歸零
    總額 += 金額
print("錯誤版本：", 總額, "← 只剩最後一筆！")
````

````python
# ⚠️ 最常見的錯誤 2：print 放錯層（縮排差一格，意義完全不同）
總額 = 0
print("--- print 在迴圈裡（印 6 次）---")
for 金額 in 今日訂單:
    總額 += 金額
    print(總額)

總額 = 0
for 金額 in 今日訂單:
    總額 += 金額
print("--- print 在迴圈外（印 1 次，通常你要的是這個）---")
print(總額)
````

### 找最大值

````python
最大 = 今日訂單[0]          # 先假設第一筆最大
for 金額 in 今日訂單:
    if 金額 > 最大:
        最大 = 金額
print("最貴：", 最大)
````

⚠️ 起始值不要寫 `最大 = 0`。萬一資料全是負數（例如退款紀錄）就錯了。
**用第一筆當起點才安全。**

````python
# 親眼看 最大 = 0 的錯誤
退款紀錄 = [-30, -65, -45]

錯的最大 = 0
for x in 退款紀錄:
    if x > 錯的最大:
        錯的最大 = x

對的最大 = 退款紀錄[0]
for x in 退款紀錄:
    if x > 對的最大:
        對的最大 = x

print("從 0 開始：", 錯的最大, "← 這筆資料根本不存在！")
print("從第一筆開始：", 對的最大, "✔")
````

> 🔍 Python 內建 `sum()`、`max()`、`len()` 直接就能做。
> **那為什麼還要手寫迴圈？** 因為內建函式只處理最單純的情況。
> 當阿宏說「只算會員的、而且排除退款」時，你還是得自己寫。**先懂原理，再用捷徑。**

````python
print(sum(今日訂單), max(今日訂單), min(今日訂單), len(今日訂單))
````

## 4.4 range()：產生一串數字

````python
print(list(range(5)))          # 0~4  ⚠️ 不含 5
print(list(range(1, 6)))       # 起點, 終點（不含）
print(list(range(0, 10, 2)))   # 起點, 終點, 間隔
print(list(range(5, 0, -1)))   # 倒數
````

⚠️ **`range(5)` 不包含 5**（半開區間）。
好處是 `range(len(清單))` 剛好對應所有合法索引，而且 `range(a, b)` 的長度剛好是 `b - a`。

````python
# enumerate()：同時要編號和內容
print("😩 笨方法")
for i in range(len(今日訂單)):
    print(i, 今日訂單[i])

print("\n🙂 Python 的方法")
for 序號, 金額 in enumerate(今日訂單, start=1):
    print(f"第 {序號} 筆：{金額} 元")
````

**看到 `range(len(...))` 就想到 `enumerate()`** —— 這是 Python 風格的重要指標。

````python
# zip()：兩個清單一起跑
品項清單 = ["珍奶", "冬瓜", "紅茶"]
價格清單 = [65, 45, 35]

for 品項, 價格 in zip(品項清單, 價格清單):
    print(f"{品項}：{價格} 元")
````

## 4.5 while：條件成立就一直做

`for` 是「已知要做幾次」，`while` 是「做到某個條件為止」。

````python
庫存 = 10
賣出 = 0
while 庫存 > 0:
    庫存 -= 1
    賣出 += 1
print(f"賣完了，共賣出 {賣出} 杯")
````

### ⚠️ 無窮迴圈

```python
庫存 = 10
while 庫存 > 0:
    print("賣一杯")     # 忘了減庫存 → 永遠停不下來
```

**Jupyter 按 `I` 兩下（或工具列 ■）中斷；終端機按 `Ctrl + C`。**

寫 `while` 養成習慣問兩個問題：

1. 迴圈裡**哪一行**會讓條件逐漸變成假？
2. 有沒有可能永遠變不成假？

````python
# 加保險絲：怕寫錯就設一個上限
次數 = 0
庫存 = 10
while 庫存 > 0:
    庫存 -= 1
    次數 += 1
    if 次數 > 1000:                # 保險絲
        raise RuntimeError("迴圈跑太多次，可能寫錯了")
print("🟢 安全結束，跑了", 次數, "圈")
````

⚠️ **`input()` 在 Jupyter 裡會跳輸入框，在自動化測試中會卡住。**
本教材的 notebook 一律不用 `input()`。
第 14 章會教你把「互動」和「邏輯」分開，好讓邏輯可以被測試 —— 這是 TDD 帶來的重要設計習慣。

## 4.6 break / continue / for-else

````python
for 金額 in 今日訂單:
    if 金額 > 100:
        print("找到大額訂單", 金額)
        break              # 立刻跳出整個迴圈
````

````python
含退款 = [65, 0, 45, 0, 120]
有效總額 = 0
for 金額 in 含退款:
    if 金額 == 0:
        continue           # 跳過這一圈，直接進下一圈
    有效總額 += 金額
print("有效總額：", 有效總額)
````

````python
# Python 獨有的 for-else：迴圈「沒有被 break」才執行 else
for 金額 in 今日訂單:
    if 金額 > 1000:
        print("有超大訂單")
        break
else:
    print("今天沒有超大訂單")       # 沒 break 過才執行
````

> 🔍 這個 `else` 語意很違反直覺（很多人認為它命名錯誤），用途是「搜尋失敗時做什麼」。
> 初學階段知道有這東西即可，用一個布林旗標也能達到同樣效果。

## 4.7 巢狀迴圈

````python
for 杯型 in ["中杯", "大杯"]:
    for 甜度 in ["無糖", "半糖", "全糖"]:
        print(f"{杯型}{甜度}", end="  ")
    print()
````

## 4.8 🟢 讓測試變綠

````python
from decimal import Decimal, ROUND_HALF_UP


def 四捨五入(數值) -> int:
    return int(Decimal(str(數值)).quantize(Decimal("1"), rounding=ROUND_HALF_UP))


def 總營收(訂單):
    總額 = 0
    for 金額 in 訂單:
        總額 += 金額
    return 總額


def 最貴一筆(訂單):
    最大 = 訂單[0]
    for 金額 in 訂單:
        if 金額 > 最大:
            最大 = 金額
    return 最大


def 筆數(訂單):
    數 = 0
    for _ in 訂單:           # _ 代表「這個值我不用」
        數 += 1
    return 數


def 平均客單價(訂單):
    return 四捨五入(總營收(訂單) / 筆數(訂單))
````

````python
assert 總營收(今日訂單) == 440
assert 最貴一筆(今日訂單) == 120
assert 筆數(今日訂單) == 6
assert 平均客單價(今日訂單) == 73
print("🟢 全綠")
````

### 🔵 重構：現在測試罩著你，可以放心改寫

````python
def 總營收(訂單):
    return sum(訂單)


def 最貴一筆(訂單):
    return max(訂單)


def 筆數(訂單):
    return len(訂單)


def 平均客單價(訂單):
    return 四捨五入(sum(訂單) / len(訂單))


# 一模一樣的測試，再跑一次
assert 總營收(今日訂單) == 440
assert 最貴一筆(今日訂單) == 120
assert 筆數(今日訂單) == 6
assert 平均客單價(今日訂單) == 73
print("🟢 重構後還是全綠 —— 證明行為沒被改壞")
````

> 🔍 **你剛剛完整體驗了 TDD 三步驟：🔴 紅 → 🟢 綠 → 🔵 重構。**
>
> 重構是「不改變行為，只改善結構」。你怎麼知道行為沒變？**因為測試還是綠的。**
> 沒有測試的重構，叫做「改東西然後祈禱」。
